```python
# data
starting_stock = 12000
monthly_doses_administered = [900, 1100, 1250, 800, 950, 1400, 1300, 1000, 850, 1150, 1200, 900]

# holding variables
current_stock = starting_stock
remaining_stock_history = []



for month, doses in enumerate(monthly_doses_administered, start=1):
    # check for stockout
    if current_stock - doses < 0:
        print(f"WARNING: Stockout will occur in Month {month}!")
        print(f"Remaining stock ({current_stock}) cannot cover requested doses ({doses}).")
        break

        #accumulate
        current_stock -= doses

    # store the remaining stock in our history list
    remaining_stock_history.append(current_stock)
    
    # print the monthly progress
    print(f"Month {month}: {current_stock} doses remaining")

# verify the bonus list after the loop finishes or breaks:
print("\nStock history list:", remaining_stock_history)     
        
    
```

### `Range()` Loop a specific number of times

When you don't have a sequence but need to repeat an action a set number of times, use range(). It generates a sequence of numbers on the fly.

• Syntax: range(start, stop, step)

• Outcome: Efficient counting without manually tracking or incrementing a variable.

```python
for i in range(3):
    print("Action executed")
```

```python
# Count from 5 to 11, skipping by 2s

for num in range(5, 12, 2):
    print(num)
```

### `enumerate()` Tracking counters safely

As seen previously, if you need the index (position) of an item while looping through a sequence, wrap your sequence in enumerate().

• Outcome: Completely eliminates the messy index = 0 and index += 1 manual tracking pattern.

```python
players = ['Mario', 'Luigi', 'Peach']

for rank, player in enumerate(players, start=1):
    print(f"Rank {rank}: {player}")
```

### `zip()` Loop through multiple loops simultaneously

If you have two or more related lists of the same length, zip() pairs their elements together side-by-side so you can loop through them in a single pass.

• Outcome: Avoids nested loops or accessing items using ugly syntax like listA[i] and listB[i].

```python
items = ['Laptop', 'Phone', 'Tablet']
prices = [1200, 800, 400]

for item, price in zip(items, prices):
    print(f"The {item} costs ${price}.")
```

### `items()` Looping through dictionaries

By default, looping over a dictionary only gives you its keys. If you want to process both the key and the value together, use the .items() method.

    • Outcome: Clean, readable key-value unpacking.

```python
stock = {"Apples": 50, "Bananas": 30}

for fruit, quantity in stock.items():
    print(f"We have {quantity} {fruit} left.")
```

### Controlling the loop: `break` and `continue`

Sometimes you need to alter the flow of the loop based on a condition inside the code block.

• `break`: Immediately terminates the entire loop and jumps outside of it.

• `continue`: Skips the rest of the current code block and moves straight to the next item in the sequence.

```python
numbers = [1, 2, 3, 4, 5]

for num in numbers:
    if num == 3:
        continue  # Skip 3 entirely
    if num == 5:
        break     # Stop the loop completely when hitting 5
    print(num)    # Outputs: 1, 2, 4
```

### List Comprehensions: Shrinking Loops into One Line

A **list comprehension** is a concise, "Pythonic" way to create a new list based on an existing sequence. It combines a for loop and an optional conditional statement into a single line of code.

**The Transformation**

Instead of creating an empty list, looping, and appending items, you build the list directly around the loop.

### Standard Way: 4 lines

```python
numbers = [1, 2, 3, 4]
squares = []

for num in numbers:
    squares.append(num * num)

for i in squares:
    print(i)
```

### List Comprehension Way: 1 line

```python
squares = [num * num for num in numbers]
squares
```

### Conditional Statements in a List Comprehension

```python
ages = [12, 21, 34, 45, 60]
evens = [age for age in ages if age % 2 == 0]
evens
```

### Nested loops: loops inside loops

A nested loop is simply a loop inside another loop.

• The outer loop triggers the inner loop.

• For every single iteration of the outer loop, the inner loop executes from start to finish.

Basic Grid Example

Think of a nested loop like reading rows and columns on a spreadsheet.

```python
for row in range(1, 3):
    for col in range(1, 4):
        print(f"Row {row}, Col {col}")
```

### Loops for modeling and simulation

#### Time stepping loop

This is the simulation loop. State evolves and each step depends on the last, so there’s no vectorizing it away.

```python
s, i, r = 999, 1, 0
n = s + i + r
beta, gamma = 0.3, 0.1
history = []

for day in range(180):
    new_inf = beta * s * i / n
    new_rec = gamma * i
    s -= new_inf
    i += new_inf - new_rec
    r += new_rec
    history.append({"day": day, "s": s, "i": i, "r": r})
```
