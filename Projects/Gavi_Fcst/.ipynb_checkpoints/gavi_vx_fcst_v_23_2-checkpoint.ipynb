# 1.1 Reading messy Excel *properly*

Beginners often do `pd.read_excel(path)` and immediately start analyzing whatever comes back.
That's a trap. Real business spreadsheets have **title banners, notes, and blank rows** above the
actual table. If you don't tell pandas where the real header row is, every column name and every
dtype will be wrong — and you won't necessarily get an error, just silently bad data.

**So step one is always: look at the raw sheet before parsing it.**

## 1.1 What sheets are in the workbook?

`pd.ExcelFile` opens the workbook *once* and lets us inspect it cheaply before reading anything heavy.

```python
xl = pd.ExcelFile(data_path)
xl.sheet_names
```

Three sheets:
- v23.2 Forecast Data (Internal) -- the main cost/volume table.
- v23.2 Key Prog Assumptions -- the coverage/population/wastage drivers.
- Pivot -- a pre-built Excel summary we'll reproduce as a self-check.

## 1.2 Peek at the raw top-left corner (no header assumptions)

We read with header=None so pandas makes no assumption about where the header is.  Then we look at the first several rows and columns to *see* the layout with our own eyes.

```python
raw = pd.read_excel(data_path, sheet_name="v23.2 Forecast Data (Internal)", header=None)
print("Raw shape (rows, cols):",raw.shape)
raw.iloc[:8, :8] # first 8 rows, first 8 columns
```

```python
raw.head()
```

## 1.3 Read the sheet with the correct header row

Now we read for real, telling pandas the header row is row 5.

```python
fc_raw = pd.read_excel(data_path, sheet_name="v23.2 Forecast Data (Internal)", header=5)
print("Shape with correct header:", fc_raw.shape)
list(fc_raw.columns)
```

```python
fc_raw.head()
```

This is better, but there are unnamed columns, there are columns with trailing spaces 'AFC LVL1 ' and misspelled 'Quantitiy in M'.

## 1.4 Column hygiene

We (a) drop the unnamed margin column and (b) strip stray whitespace from every column name.  We deliberately **keep** the original misspelling for now so you can see it, then decide what to do.

```python
fc = fc_raw.copy()

# (a) Drop any fully-unnamed columns (the empty Excel margin).
fc = fc.loc[:, ~fc.columns.astype(str).str.startswith("Unnamed")]

# (b) Strip leading/trailing whitespace from all column names.
fc.columns = fc.columns.str.strip()

list(fc.columns)
```

# Part 2 Systematic Exploratory Data Analysis (EDA)

EDA is where you *earn the right* to trust your data.  We'll go through a repeatable checklist you can reuse on any dataset:
1. **Shape & a glance** -- how big, what do the rows look like?
2. **Types** -- is each column the dtype it should be?
3. **Missingness** -- where are the nulls, and *why*?
4. **Cardinality** -- how many distinct values per categorical column?
5. **Constants** -- columns with only one value (candidates to drop).
6. **A domain surprise** -- the "Antigen" column is not what it seems.
7. **Integrity checks** -- do the columns agree with each other?

Treat this as a *template*, not a one-off.

## 2.1 Shape and a glance

```python
print("Rows, Columns:", fc.shape)
fc.head()
```

```python
# .info() is the fastest single view: column names, non-null counts, and dtypes together.
fc.info()
```

## 2.2 Summary statistics

`.describe()` on numeric columns gives count/mean/std/min/quartiles/max.  On categorical columns (include="object") it gives count/unique/top/frequency.  Both are worth a look.

```python
fc.describe()
```

```python
fc.describe(include="object")
```

Two things already jump out:

- Several object columns have unique = 1 (e.g. the top value equals every value).  Those are constant columns; see 2.5.

## 2.3 Missingness -- and *why*

A null count is a question, not an answer.  The analyst's job is to explain *why* values are missing, because the reason dictates what you're allowed to do about them.

```python
# Null count per column, largest first.
fc.isna().sum().sort_values(ascending=False)
```

```python
# Only Quantity has nulls.  Let's LOOK at those 93 rows instead of guessing.
missing_q = fc[fc["Quantity"].isna()]
print("Rows with missing Quantity:", len(missing_q))
missing_q[["Antigen", "Support Type", "Category", "USD", "Quantity", "Quantitiy in M"]].head(10)
```

```python
# Among rows where Quantity is NaN, what values does "Quantitiy in M' take?
missing_q["Quantitiy in M"].value_counts(dropna=False)
```

## 2.4 Cardinality -- how many distinct values per categorical column?

Before analyzing categories, know how many there are.  This shapes every groupby decision downstream.

```python
cat_cols = fc.select_dtypes(include="object").columns
fc[cat_cols].nunique().sort_values()
```

## 2.5 Constant columns (only one distinct value)

Columns with a single value carry **no information** for analysis (they can't distinguish rows).  They're usually metadata: which forecast version, which period.  We identify them programmatically rather than by eyeballing.

```python
constant_cols = [c for c in fc.columns if fc[c].nunique(dropna=False) == 1]
print("Constant columns:", constant_cols)
for c in constant_cols:
    print(f"  {c!r:14s} -> only value: {fc[c].iloc[0]!r}")
```

## 2.6 The domain surprise: "Antigen" is a mixed bag

This is the most important discovery in the whole EDA.  Let's list every distinct Antigen.

```python
sorted(fc["Antigen"].unique())
```

This demonstrates some financial mechanisms and offsets.  Some of these are not dose driven like those items that are truly antigens.

```python
# For each Antigen, the fraction of its rows that have positive doses.

dose_share = (
    fc.assign(has_doses=fc["Quantitiy in M"] > 0)
    .groupby("Antigen")["has_doses"]
    .mean()
    .sort_values()
)

never_doses = dose_share[dose_share == 0].index.tolist() # pure financial line items
always_doses = dose_share[dose_share == 1].index.tolist() # always real product
mixed_doses = dose_share[(dose_share > 0) & (dose_share < 1)].index.tolist()

print("NEVER any doses (financial / adjustment lines):")
print("  ", never_doses)
print("\nALWAYS has doses (clearly real vaccine product):")
print("  ", always_doses)
print("\nMIXED (real vaccine in some rows, adjustment in others):")
print("  ", mixed_doses)
```

```python
financial_rows = fc[fc["Antigen"].isin(never_doses)]
print("Financial line-item rows:", len(financial_rows))
print("Their total USD in M:", round(financial_rows["USD in M"].sum(), 2))
print("\nBiggest negative entries (savings / downward adjustments):")
financial_rows.sort_values("USD in M").head(6)[["Antigen","Support Type","Fund Center","Year","USD in M"]]
```

## 2.7 Integrity Checks--do the columns agree with each other? 

The file gives us USD and USD in M, plus Quantity and Quantitiy in M.  If "in M" means "in millions," then USD / 1,000,000 should equal USD in M.  **Never assume--test it.**  These checks catch corrupted exports and copy-paste errors before they reach your conclusions.

```python
usd_diff = (fc["USD"] / 1e6 - fc["USD in M"]).abs().max()
qty_diff = (fc["Quantity"] / 1e6 - fc["Quantitiy in M"]).abs().max()

print(f"Max discrepancy USD/1e6      vs USD in M       : {usd_diff:.3e}")
print(f"Max discrepancy Quantity/1e6 vs Quantitiy in M : {qty_diff:.3e}")
print("\nInterpretation: both ~0  -> the '... in M' columns are exact scaled copies (redundant).")
```

```python
# A reusable boolean flag: is this row a real vaccine product line (has doses) or a financial adjustment?
fc["is_financial_line"] = fc["Antigen"].isin(never_doses)

# Convenience view: real product only. We'll use this for the vaccine analyses in Part 3.
vax = fc[~fc["is_financial_line"]].copy()
print("All rows:", len(fc), "| Real-product rows:", len(vax), "| Financial-line rows:", fc['is_financial_line'].sum())
```

# Part 3 Cost & volume by antigen, country, and year

With a trustworthy frame, aggregation is the fun part.  We'll build up from a single **groupby** to a `pivot_table`, add a derived metric (price per dose), and make a couple of clean charts.  Throughout, we work from vax (real product only) unless we explictly want the financial lines.

## 3.1 Total cost by antigen (and reproduce the Excel Pivot as a self-check)

The single most valuable habit: when a "ground truth" exists, reproduce it and confirm you match.  The `Pivot` she sums USD in M by antigen across **all** rows (including the financial lines), so for an apples-to-apples comparison we use the *full* fc, not vax.

```python
cost_by_antigen = (
    fc.groupby("Antigen")["USD in M"]
    .sum()
    .sort_values(ascending=False)
)
cost_by_antigen.head(10)
```

```python
# Load the Excel Pivot sheet (its header is on row index 2) and compare our numbers to it.

pivot_sheet = (
    pd.read_excel(data_path, sheet_name="Pivot", header=2)
        .dropna(how="all")
        .rename(columns={"Row Labels": "Antigen", "Sum of USD in M": "USD_in_M_excel"})
)

# Excel's own grand total row - pull it out for a headline comparison.

excel_grand_total = pivot_sheet.loc[pivot_sheet["Antigen"] == "Grand Total", "USD_in_M_excel"].iloc[0]
our_grand_total = cost_by_antigen.sum()

print(f"Our grand total   : {our_grand_total:,.4f} (M USD)")
print(f"Excel grand total : {excel_grand_total:,.4f} (M USD)")
print(f"Difference        : {abs(our_grand_total - excel_grand_total):.6f}")
```

## 3.2 A quick, clean bar chart of top antigens by cost

```python
top10 = cost_by_antigen.head(10)

fig, ax = plt.subplots(figsize=(9, 5))
ax.barh(top10.index[::-1], top10.values[::-1])    # reverse so the biggest bar is on top
ax.set_xlabel("Forecast cost, 2026-2030 (millions USD)")
ax.set_title("Top 10 antigens by total forecast procurement cost (v23.2)")
for i, v in enumerate(top10.values[::-1]):
    ax.text(v, i, f" {v:,.0f}", va="center")
plt.tight_layout()
plt.show()
```

## 3.3 Cost trend over time (real product only)

How does forecasted vaccine spend evolve across 2026-2030? Here we deliberately use vax to exclude the financial adjustment lines, so the trend reflects actual product.

```python
cost_by_year = vax.groupby("Year")["USD in M"].sum()
print(cost_by_year)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(cost_by_year.index, cost_by_year.values, marker="o")
ax.set_xlabel("Year"); ax.set_ylabel("Cost (millions USD)")
ax.set_title("Forecast vaccine procurement cost by year (real product)")
ax.set_xticks(cost_by_year.index)
ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

## 3.4 Two dimensions at once: `pivot_table` (antigen x years)

`groupby` with one key gives a list.  When you want a **matrix**--rows=antigen,columns=year--reach for `pivot_table`.  This is the pandas equivalent of an Excel PivotTable.

```python
antigen_year = pd.pivot_table(
    vax,
    index="Antigen",
    columns="Year",
    values="USD in M",
    aggfunc="sum",
    margins=True,
    margins_name="Total",
).sort_values("Total", ascending=False)

antigen_year.head(12)
```

## 3.6 A derived metric: implied price per dose

Cost and volume are more insightful *together*.  Cost per dose = USD $\div$ doses.  We compute it at the antigen level (summing each first, then dividing--**not** averaging per-row ratios, which would overweight tiny rows).

```python
per_antigen = (
    vax.groupby("Antigen")
        .agg(total_usd=("USD", "sum"), total_doses=("Quantity", "sum"))
)

per_antigen["usd_per_dose"] = per_antigen["total_usd"] / per_antigen["total_doses"]
per_antigen.sort_values("usd_per_dose", ascending=False).round(3).head(12)
```

# Part 4 Linking the forecast to its assumptions

The payoff: the second sheet tells us *why* the numbers look the way they do--population, coverage, wastage, buffer.  But joining the two sheets is genuinely hard because **they don't share clean keys**.  This is the real world, and learning to handle it is a core skill.

We'll (1) load and clean the assumptions sheet, (2) diagnose the key mismatches, (3) build a small, honest crosswalk, and (4) perform a *validated* merge--checking how many rows actually matched rather than trusting the join blindly.

## 4.1 Load and clean the assumptions sheet

Its real header is on row index 2.  Same defensive read as before.

```python
ka_raw = pd.read_excel(data_path, sheet_name="v23.2 Key Prog Assumptions", header=2)
ka = ka_raw.loc[:, ~ka_raw.columns.astype(str).str.startswith("Unnamed")].copy()
ka.columns = ka.columns.str.strip()
print("Assumptions shape:", ka.shape)
ka.head()
```

Columns of interest: ISO3 (country code), VaccineGroup, SupportName, PolicyName, Year, Population_forecasted, Cov1..Cov4_forecasted (dose-1 through dose-4 coverage), Wastage_forecasted, Buffer_policy_forecasted.

## 4.2 Diagnose the key mismatch

Put the two sets of keys side by side and see where they *fail* to line up.

```python
print("Forecast 'Antigen' values:")
print(" ", sorted(vax["Antigen"].unique()))
print("\nAssumptions 'VaccineGroup' values:")
print(" ", sorted(ka["VaccineGroup"].unique()))

print("\nForecast 'Fund Center' (country NAMES), sample:")
print(" ", sorted(vax["Fund Center"].unique())[:8], "...")
print("\nAssumptions 'ISO3' (country CODES), sample:")
print(" ", sorted(ka["ISO3"].unique())[:8], "...")
```

4.3 Build an antigen to VaccineGroup crosswalk

```python
antigen_to_group = {
    "PENTA": "Pentavalent",
    "MAL": "Malaria",
    "RV": "Rota",
    "HEXA": "Hexa",
    "Rabies PEP": "Rabies",
    "MR": "MR",
    "PCV": "PCV",
    "IPV": "IPV",
    "HPV": "HPV",
    "YF": "YF",
    "MenA": "MenA",
    "OCV": "OCV",
    "Measles": "Measles",
    "TCV": "TCV",
    "JEV": "JEV",
    "HepB": "HepB",
    "MMCV": "MMCV",
    "RSV": "RSV",
}

vax_antigens = set(vax["Antigen"].unique())
mapped   = sorted(a for a in vax_antigens if a in antigen_to_group)
unmapped = sorted(a for a in vax_antigens if a not in antigen_to_group)
print(f"Mapped {len(mapped)} antigens; {len(unmapped)} left unmapped (no confident match).")
print("Unmapped:", unmapped)
```

Better to leave these unmapped and to ask an SME how to map them.

## 4.4 Build a country name to ISO3 crosswalk

There's no ISO3 in the forecast sheet, so we need an external mapping.  The tidy way is the `pycountry` library, but it may not be installed--so we try it, and fall back to building the crosswalk **from the data itself** for the countries we can match by fuzzy name.  For this teaching example we'll take the robust path: derive the crosswalk with `pycountry` if available, else map the overlap we can.

```python
name_to_iso3 = {}
try:
    import pycountry
    for name in vax["Fund Center"].unique():
        match = None
        try:
            match = pycountry.countries.lookup(name)
        except LookupError:
            # try a fuzzy search for tricky names
            try:
                match = pycountry.countries.search_fuzzy(name)[0]
            except Exception:
                match = None
        if match is not None:
            name_to_iso3[name] = match.alpha_3
    source = "pycountry"
except ImportError:
    # Minimal fallback so the notebook still runs without the library installed.
    # (Extend this dict as needed; or run:  pip install pycountry  )
    name_to_iso3 = {
        "Afghanistan": "AFG", "Bangladesh": "BGD", "Ethiopia": "ETH", "Kenya": "KEN",
        "Nigeria": "NGA", "Pakistan": "PAK", "Uganda": "UGA", "Mozambique": "MOZ",
        "Democratic Republic of the Congo": "COD", "Tanzania": "TZA", "India": "IND",
    }
    source = "fallback dict (install pycountry for full coverage)"

print(f"Country crosswalk source: {source}")
print(f"Mapped {len(name_to_iso3)} of {vax['Fund Center'].nunique()} country names to ISO3.")
missing = sorted(set(vax["Fund Center"].unique()) - set(name_to_iso3))
print("Unmapped countries (first 10):", missing[:10])
```

## 4.5 Attach the crosswalk keys to the forecast, then aggregate to the join grain

The assumptions are at **country** $\times$ **vaccine** $\times$ **support** $\times$ **policy** $\times$ **year**, while our forecast row grain also splits into `Category`.  To join cleanly we first **aggregate the forecast** to `ISO3` $\times$ `VaccineGroup` $\times$ `Year` (summing across categories and support types), which matches a level the assumptions can speak to.

```python
vax_keyed = vax.copy()
vax_keyed["VaccineGroup"] = vax_keyed["Antigen"].map(antigen_to_group)
vax_keyed["ISO3"] = vax_keyed["Fund Center"].map(name_to_iso3)

# Keep only rows where BOTH keys resolved — be explicit about what we're dropping and why.
before = len(vax_keyed)
vax_keyed = vax_keyed.dropna(subset=["VaccineGroup", "ISO3"])
print(f"Rows retained for join: {len(vax_keyed)} of {before} "
      f"({len(vax_keyed)/before:.0%}) — the rest lacked a confident key mapping.")

forecast_grain = (
    vax_keyed.groupby(["ISO3", "VaccineGroup", "Year"], as_index=False)
             .agg(usd_m=("USD in M", "sum"), doses_m=("Quantitiy in M", "sum"))
)
forecast_grain.head()
```

## 4.6 Aggregate the assumptions to the same grain, then merge--*with validation*

A vaccine can appear under several `SupportName`/`PolicyName` rows in the assumptions.  For a first pass we take the routine, standard-policy assumption per country/vaccine/year.  Then we merge and--crucially--use `indicator=True` to *see* how many keys matched on each side instead of assuming.

```python
assump = ka[(ka["SupportName"] == "Routine") & (ka["PolicyName"] == "Standard")].copy()
assump_grain = (
    assump.groupby(["ISO3", "VaccineGroup", "Year"], as_index=False)
          .agg(population=("Population_forecasted", "sum"),
               cov1=("Cov1_forecasted", "mean"),
               cov3=("Cov3_forecasted", "mean"),
               wastage=("Wastage_forecasted", "mean"))
)

merged = forecast_grain.merge(
    assump_grain,
    on=["ISO3", "VaccineGroup", "Year"],
    how="left",             # keep every forecast row; attach assumptions where they exist
    indicator=True,         # adds a '_merge' column telling us what matched
)

print("Match diagnostics (how many forecast rows found assumptions):")
print(merged["_merge"].value_counts())
print(f"\nMatch rate: {(merged['_merge'] == 'both').mean():.0%}")
merged.head()
```

## 4.7 A worked sanity check: do doses roughly track population $\times$ coverage?

For rows that matched, a crude expectation is: doses $\approx$ target population $\times$ coverage ($\times$ a wastage factor).  We won't nail it exactly (buffers, multi-dose schedules, campaigns), but the *relationship* should be visible--and checking it builds intuition for what the assumptions actually do.

```python
check = merged[merged["_merge"] == "both"].copy()
check["expected_doses_m"] = (check["population"] * check["cov1"]) / 1e6  # very rough, dose-1 only

# Correlation between forecast doses and a population*coverage expectation.
corr = check[["doses_m", "expected_doses_m"]].corr().iloc[0, 1]
print(f"Correlation (forecast doses vs population*coverage): {corr:.2f}")

fig, ax = plt.subplots(figsize=(6.5, 6.5))
ax.scatter(check["expected_doses_m"], check["doses_m"], alpha=0.5)
lims = [0, max(check["expected_doses_m"].max(), check["doses_m"].max())]
ax.plot(lims, lims, linestyle="--")   # y = x reference line
ax.set_xlabel("Population × Cov1  (millions)")
ax.set_ylabel("Forecast doses (millions)")
ax.set_title("Do forecast doses track population × coverage?")
plt.tight_layout(); plt.show()
```
