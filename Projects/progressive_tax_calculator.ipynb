```python
# define the tax brackets first in a dictionary differentiating between single and married filing jointly, then tuples for each cap and rate

tax_brackets_2027 = {
    "single": {
        "standard_deduction": 16600,
        "brackets": [
            (12800, 0.10),
            (52025, 0.12),
            (109125, 0.22),
            (208325, 0.24),
            (264550, 0.32),
            (661375, 0.35),
            (float("inf"), 0.37),
        ],
    },
    "married_joint": {
        "standard_deduction": 29200,
        "brackets": [
            (25600, 0.10),
            (104050, 0.12),
            (218250, 0.22),
            (416650, 0.24),
            (529100, 0.32),
            (793650, 0.35),
            (float("inf"), 0.37),
        ],
    },
}
```

```python
# define the function

def calculate_tax(gross_income, filing_status="single"):
    if filing_status not in tax_brackets_2027:
        raise ValueError(f"Unknown status: {filing_status}. Choose 'single' or 'married_joint'.")

    status_data = tax_brackets_2027[filing_status]
    deduction = status_data["standard_deduction"]
    taxable_income = max(0.0, gross_income - deduction)

    if taxable_income == 0:
        return 0.0

    total_tax = 0.0
    previous_threshold = 0

    for threshold, rate in status_data["brackets"]:
        if taxable_income > previous_threshold:
            taxable_amount_in_bracket = min(taxable_income, threshold) - previous_threshold
            total_tax += taxable_amount_in_bracket * rate
            previous_threshold = threshold
        else:
            break

    return total_tax
```

```python
# run the function

estimated_tax_2027 = calculate_tax(344580.3, "single")
est_per_check_2027 = estimated_tax_2027 / 24
current_tax_check = 4684.68
difference_2026_2027 = current_tax_check - est_per_check_2027

print(f"Your estimated tax for 2027 is ${estimated_tax_2027:,.2f}.")
print(f"Your estimated tax per check is ${est_per_check_2027:,.2f}.")
print(f"This is a diffference of ${difference_2026_2027:,.2f}.")
```
