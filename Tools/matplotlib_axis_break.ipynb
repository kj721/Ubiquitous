```python
import matplotlib.pyplot as plt

years = [2019, 2020, 2021, 2022, 2023, 2024, 2025]
values = [20, 22, 95, 44, 34, 45, 90]

fig, (ax_top, ax_bottom) = plt.subplots(2, 1, sharex=True, figsize=(8, 6))

# plot the same data on both axes
ax_top.plot(years, values)
ax_bottom.plot(years, values)

# each axis only shows part of the y-range
ax_top.set_ylim(80, 100)      # zoom into the high range
ax_bottom.set_ylim(0, 20)     # zoom into the low range

# hide the spines between them
ax_top.spines['bottom'].set_visible(False)
ax_bottom.spines['top'].set_visible(False)
ax_top.tick_params(bottom=False)

# add small diagonal "break" marks
d = 0.5
kwargs = dict(marker=[(-1, -d), (1, d)], markersize=12,
              linestyle="none", color='k', mec='k', mew=1, clip_on=False)
ax_top.plot([0], [0], transform=ax_top.transAxes, **kwargs)
ax_bottom.plot([0], [1], transform=ax_bottom.transAxes, **kwargs)

plt.tight_layout()
plt.show()
```
